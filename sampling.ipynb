## Sampling Project

📌 1. Introduction
Research Question:
What are the streaming platform preferences and usage patterns among university students?

Target Population:
All undergraduate students at a medium-sized urban university.

Chosen Sampling Method:
Stratified Sampling — to ensure fair representation from each academic year (Freshman, Sophomore, Junior, Senior).

🧪 2. Methodology
Sampling Technique:
The student population is divided into four strata: Freshman, Sophomore, Junior, and Senior.

From each stratum, 5 students were randomly selected using a name generator.

Sample Size:
20 participants
(5 per academic year)

Data Collection:
Questionnaire with 6 multiple-choice and short-answer questions distributed via Google Form.

### Data generation:

```python
import pandas as pd
import random

# Set random seed for reproducibility
random.seed(42)

# Define possible responses
years = ["Freshman", "Sophomore", "Junior", "Senior"]
streaming_services = ["Netflix", "Disney+", "Prime Video", "Hulu", "Other"]
devices = ["Laptop", "Smartphone", "Smart TV", "Tablet"]
content_types = ["TV Series", "Movies", "Documentaries", "Reality TV", "Anime"]
satisfaction_scores = [1, 2, 3, 4, 5]

# Generate data for 20 participants
data = []
for i in range(20):
    year = years[i % 4]  # Ensure 5 from each academic year
    platform = random.choices(streaming_services, weights=[40, 20, 15, 15, 10])[0]
    hours = random.randint(3, 22)
    device = random.choices(devices, weights=[50, 30, 15, 5])[0]
    shares_account = random.choices(["Yes", "No"], weights=[70, 30])[0]
    content = random.choices(content_types, weights=[50, 25, 10, 10, 5])[0]
    satisfaction = random.choices(satisfaction_scores, weights=[5, 10, 15, 30, 40])[0]
    
    data.append([
        i+1,
        year,
        platform,
        hours,
        device,
        shares_account,
        content,
        satisfaction
    ])

# Create DataFrame
columns = [
    "Participant_ID",
    "Academic_Year",
    "Most_Used_Platform",
    "Hours_Per_Week",
    "Primary_Device",
    "Shares_Account",
    "Preferred_Content_Type",
    "Satisfaction_Score"
]
df = pd.DataFrame(data, columns=columns)
df.head()
```
