## 🧪 Sampling Error Analysis

### ⚠️ Identified Sources of Sampling Error

1. **Non-representative Sample**  
   The dataset was sourced from Kaggle and may not be representative of the broader population of book consumers in Germany. For example, it is unclear how participants were selected and whether diverse age groups, regions, or education levels were proportionally included.

2. **Sampling Bias from Convenience or Voluntary Response**  
   The original dataset may suffer from **self-selection bias**, meaning that only individuals who are particularly engaged with reading or surveys might have chosen to participate.

3. **Limited Sample Size**  
   With only **64 observations**, statistical conclusions have a higher degree of uncertainty. Smaller samples are more vulnerable to random fluctuations and less likely to reflect true population parameters.

---

### 🧮 Calculation of Margin of Error

To estimate the precision of our sample mean for **Weekly Reading Time**, we calculated the **margin of error** for a 95% confidence level:

- **Sample Mean Reading Time:** 4.65 hours  
- **Margin of Error (±):** 0.66 hours  
- **95% Confidence Interval (CI):** [**3.99**, **5.31**]  

This means we can be 95% confident that the true average weekly reading time in the entire population lies between **3.99 and 5.31 hours**.

---

### 💭 Impact on Conclusions

- The margin of error shows that our estimate of the population's reading time is **not exact**, but rather falls within a range.
- Sampling errors like non-representative gender or genre distributions could **skew mood correlations or genre preferences**, impacting insights and business recommendations.
- Misestimating reading habits could lead to **flawed marketing strategies** or **product targeting** in the book industry.

---

### ✅ Strategies to Reduce Sampling Error in Future Studies

1. **Increase Sample Size**  
   Collect more data to reduce the standard error and narrow the confidence interval.

2. **Use Probabilistic Sampling Methods**  
   Whenever possible, apply techniques like **stratified random sampling** across additional variables (e.g., region, age, education).

3. **Define and Control the Sampling Frame**  
   Ensure the dataset covers the full population of interest — in this case, German book readers across diverse demographics.

4. **Transparency in Data Collection**  
   Use or generate datasets with well-documented sampling procedures to assess potential bias and generalizability more effectively.

```python
categorical_columns = ['Gender', 'Favorite_Book_Genre', 'Mood_Impact']

for col in categorical_columns:
    plt.figure(figsize=(6, 4))
    ax = sns.countplot(data=df, x=col, palette='pastel',legend=False)

    total = len(df[col])

# describe bars with %
    for p in ax.patches:
        count = int(p.get_height())
        percentage = 100 * count / total
        ax.annotate(f'{percentage:.1f}%',
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='bottom', fontsize=10, color='black')

    plt.title(f"{col} – Distribution in %")
    plt.xlabel(col)
    plt.ylabel("count")
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
# Visualisiations

# average reading hours by mood
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='Mood_Impact', y='Weekly_Reading_Time(hours)', palette='Set2', estimator=np.mean, hue='Mood_Impact')
plt.title("average reading time by mood")
plt.xlabel("mood")
plt.ylabel("average reading time (h)")
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# boxplot reading time by gender
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='Gender', y='Weekly_Reading_Time(hours)', palette='pastel')
plt.title("reading time per week by gender")
plt.xlabel("gender")
plt.ylabel("reading time (h)")
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# tranform mood categories to numeric values
mood_mapping = {'Negative': -1, 'Neutral': 0, 'Positive': 1}
df['Mood_Numeric'] = df['Mood_Impact'].map(mood_mapping)

# Scatterplot
plt.figure(figsize=(6, 4))
sns.stripplot(data=df, x='Mood_Numeric', y='Weekly_Reading_Time(hours)', jitter=True, palette='coolwarm')
plt.title("Correlation between mood and reading time")
plt.xlabel("mood (-1 = negativ, 0 = neutral, 1 = positiv)")
plt.ylabel("reading time (h)")
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# correlation coefficient
correlation = df['Weekly_Reading_Time(hours)'].corr(df['Mood_Numeric'])
print(f"Correlation coefficient: {correlation:.2f}")
```

```python
# Margin of Error
# since we do not know the exact population we calculate the margin of error to predict how big the samping error
# the difference between the sampling paramerter and the paramter in the population) might be.

# Average reading time
mean = df["Weekly_Reading_Time(hours)"].mean()
std = df["Weekly_Reading_Time(hours)"].std()
n = len(df)

# std error
se = std / np.sqrt(n)

# z-value for 95%
z = stats.norm.ppf(0.975)

# Margin of Error
moe = z * se
ci_lower = mean - moe
ci_upper = mean + moe

print(f"Mean Reading Time: {mean:.2f}")
print(f"Margin of error: {moe:.2f}")
print(f"95% CI: {ci_lower:.2f} – {ci_upper:.2f})")
print(f"Based on the sample, with a chance of 95% the average reading time in the whole population lies between {ci_lower:.2f} and {ci_upper:.2f} hours per week")
```
