# Sampling Methodology Documentation

## 🎯 Chosen Sampling Method: **Stratified Random Sampling**

### Justification for Method Selection

**Why Stratified Sampling?**
1. **Representativeness**: Ensures all key demographic subgroups are represented
2. **Precision**: Reduces sampling error for subgroup analyses
3. **Comparative Analysis**: Enables valid comparisons across demographic strata
4. **Practicality**: Feasible with available demographic information

**Alternative Methods Considered:**
- **Simple Random Sampling**: Risk of under-representing minority demographic groups
- **Convenience Sampling**: High risk of selection bias
- **Cluster Sampling**: Less efficient for our target population distribution

### Sampling Plan Implementation

#### Population Definition
- **Target Population**: Urban adults aged 25-55
- **Sampling Frame**: 150 simulated metropolitan residents
- **Strata Variables**: Age, Education, Income (3×4×4 = 48 possible strata)

#### Stratification Variables
1. **Age Groups**: 25-34, 35-44, 45-55 years
2. **Education Levels**: High school or less, Some college, Bachelor's, Graduate
3. **Income Brackets**: Under $50K, $50K-$100K, $100K-$150K, $150K+

#### Sampling Procedure
1. **Frame Creation**: Generated 150 potential participants with realistic demographics
2. **Proportional Allocation**: Sample size per stratum proportional to population size
3. **Random Selection**: Simple random sampling within each stratum
4. **Data Collection**: Simulated survey administration with realistic response patterns

### Data Collection Results

#### Sample Characteristics
- **Total Participants**: 15 completed surveys
- **Age Distribution**: Proportional representation across age groups
- **Education Spread**: All education levels represented
- **Income Range**: Coverage across all income brackets

#### Response Metrics
- **Response Rate**: 100% (simulated environment)
- **Data Quality**: Complete responses for all participants
- **Representativeness**: Good coverage across all stratification variables

### Methodological Strengths

✅ **Proportional Representation**: Each demographic group represented according to population prevalence

✅ **Reduced Sampling Bias**: Systematic approach minimizes selection bias

✅ **Analytical Power**: Enables robust subgroup comparisons

✅ **Transparent Process**: Clear documentation of sampling procedure

### Limitations & Considerations

⚠️ **Simulated Data**: Responses generated algorithmically rather than real surveys

⚠️ **Limited Sample Size**: 15 participants provides initial data but limited statistical power

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set random seed for reproducibility
np.random.seed(42)
random.seed(42)

class RacialToleranceSampling:
    """
    Implementation of stratified random sampling for racial tolerance survey
    """
    
    def __init__(self):
        self.survey_data = []
        self.demographic_categories = {
            'age_groups': ['25-34', '35-44', '45-55'],
            'education_levels': ['High school or less', 'Some college', 'Bachelor\'s degree', 'Graduate degree'],
            'income_brackets': ['Under $50K', '$50K-$100K', '$100K-$150K', '$150K+']
        }
        
        # Ensure directories exist
        self.ensure_directories()
    
    def ensure_directories(self):
        """Create necessary directories if they don't exist"""
        directories = ['outputs', 'data', 'reports']
        for directory in directories:
            if not os.path.exists(directory):
                os.makedirs(directory)
                print(f"✓ Created directory: {directory}/")
    
    def generate_sampling_frame(self, population_size=200):
        """
        Create a simulated sampling frame of potential participants
        """
        print("="*60)
        print("CREATING SAMPLING FRAME")
        print("="*60)
        
        sampling_frame = []
        
        # Generate realistic demographic distribution
        age_dist = [0.4, 0.35, 0.25]  # More younger participants
        education_dist = [0.2, 0.3, 0.35, 0.15]  # Higher education bias
        income_dist = [0.25, 0.35, 0.25, 0.15]  # Realistic income distribution
        
        for i in range(population_size):
            participant = {
                'id': f"P{1000 + i}",
                'age_group': np.random.choice(self.demographic_categories['age_groups'], p=age_dist),
                'education': np.random.choice(self.demographic_categories['education_levels'], p=education_dist),
                'income': np.random.choice(self.demographic_categories['income_brackets'], p=income_dist),
                'neighborhood': np.random.choice(['Urban', 'Suburban']),
                'contacted': False,
                'responded': False
            }
            sampling_frame.append(participant)
        
        self.sampling_frame = pd.DataFrame(sampling_frame)
        print(f"✓ Created sampling frame with {len(sampling_frame)} potential participants")
        print("\nDemographic Distribution:")
        print(f"Age Groups: {self.sampling_frame['age_group'].value_counts().to_dict()}")
        print(f"Education: {self.sampling_frame['education'].value_counts().to_dict()}")
        print(f"Income: {self.sampling_frame['income'].value_counts().to_dict()}")
        
        return self.sampling_frame
    
    def stratified_sampling(self, target_sample_size=20):
        """
        Implement stratified random sampling by age, education, and income
        """
        print("\n" + "="*60)
        print("IMPLEMENTING STRATIFIED RANDOM SAMPLING")
        print("="*60)
        
        # Calculate proportional allocation
        strata_sizes = {}
        for age in self.demographic_categories['age_groups']:
            for edu in self.demographic_categories['education_levels']:
                for inc in self.demographic_categories['income_brackets']:
                    stratum = self.sampling_frame[
                        (self.sampling_frame['age_group'] == age) &
                        (self.sampling_frame['education'] == edu) &
                        (self.sampling_frame['income'] == inc)
                    ]
                    strata_sizes[(age, edu, inc)] = len(stratum)
        
        total_population = len(self.sampling_frame)
        selected_participants = []
        
        print("Sampling Strategy:")
        print("-" * 40)
        
        # Proportional allocation sampling
        for stratum_key, stratum_size in strata_sizes.items():
            if stratum_size > 0:
                age, edu, inc = stratum_key
                stratum_sample_size = max(1, round((stratum_size / total_population) * target_sample_size))
                
                stratum_participants = self.sampling_frame[
                    (self.sampling_frame['age_group'] == age) &
                    (self.sampling_frame['education'] == edu) &
                    (self.sampling_frame['income'] == inc)
                ]
                
                if len(stratum_participants) >= stratum_sample_size:
                    selected = stratum_participants.sample(n=stratum_sample_size, random_state=42)
                    selected_participants.append(selected)
                    
                    print(f"Stratum {age}/{edu}/{inc}: {stratum_sample_size} participants")
        
        self.selected_sample = pd.concat(selected_participants)
        print(f"\n✓ Selected {len(self.selected_sample)} participants through stratified sampling")
        
        return self.selected_sample
    
    def conduct_survey(self, participant):
        """
        Simulate conducting the racial tolerance survey with a participant
        """
        # Base tolerance level influenced by demographics
        base_tolerance = {
            '25-34': 3.8, '35-44': 3.5, '45-55': 3.2  # Younger more tolerant
        }
        
        education_effect = {
            'High school or less': -0.3,
            'Some college': 0.0,
            'Bachelor\'s degree': 0.2,
            'Graduate degree': 0.4
        }
        
        income_effect = {
            'Under $50K': -0.2,
            '$50K-$100K': 0.0,
            '$100K-$150K': 0.1,
            '$150K+': 0.3
        }
        
        # Calculate base score
        base_score = base_tolerance[participant['age_group']]
        base_score += education_effect[participant['education']]
        base_score += income_effect[participant['income']]
        
        # Generate survey responses with some randomness
        responses = {}
        
        # Section B: Tolerance Attitudes (1-5 Likert scale)
        for i in range(6, 11):  # Questions 6-10
            # Add individual variation around base score
            response = max(1, min(5, round(base_score + np.random.normal(0, 0.5))))
            responses[f'Q{i}'] = int(response)
        
        # Section C: Behavioral Intentions
        behavior_base = base_score - 0.5  # Behavior often lags attitudes
        responses['Q11'] = int(max(1, min(5, round(behavior_base + np.random.normal(0, 0.7)))))
        responses['Q12'] = int(max(1, min(5, round(base_score + np.random.normal(0, 0.3)))))
        responses['Q13'] = int(max(1, min(5, round(behavior_base + np.random.normal(0, 0.6)))))
        
        # Section D: Awareness
        awareness = np.random.normal(3.5, 0.8)  # Moderate awareness generally
        responses['Q14'] = int(max(1, min(5, round(awareness))))
        responses['Q15'] = int(max(1, min(5, round(np.random.normal(2.5, 1.0)))))  # Personal impact
        
        return responses
    
    def collect_data(self, sample_size=15):
        """
        Main method to implement sampling and data collection
        """
        print("\n" + "="*60)
        print("DATA COLLECTION PROCESS")
        print("="*60)
        
        # Step 1: Create sampling frame
        self.generate_sampling_frame(population_size=150)
        
        # Step 2: Implement stratified sampling
        selected_sample = self.stratified_sampling(target_sample_size=sample_size)
        
        # Step 3: Collect survey data
        print(f"\nConducting surveys with {len(selected_sample)} participants...")
        
        all_responses = []
        for idx, participant in selected_sample.iterrows():
            print(f"📝 Surveying Participant {participant['id']} "
                  f"({participant['age_group']}, {participant['education']}, {participant['income']})")
            
            # Conduct survey
            survey_responses = self.conduct_survey(participant)
            
            # Combine demographic and survey data
            complete_response = {**participant.to_dict(), **survey_responses}
            complete_response['timestamp'] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            complete_response['response_id'] = f"R{1000 + len(all_responses)}"
            
            all_responses.append(complete_response)
        
        # Create final dataset
        self.collected_data = pd.DataFrame(all_responses)
        
        print(f"\n🎉 DATA COLLECTION COMPLETE!")
        print(f"✓ Collected {len(self.collected_data)} complete surveys")
        print(f"✓ Response rate: {(len(self.collected_data)/len(selected_sample))*100:.1f}%")
        
        return self.collected_data
    
    def analyze_sample_characteristics(self):
        """
        Analyze the characteristics of the collected sample
        """
        print("\n" + "="*60)
        print("SAMPLE CHARACTERISTICS ANALYSIS")
        print("="*60)
        
        if not hasattr(self, 'collected_data'):
            print("No data collected yet. Run collect_data() first.")
            return
        
        data = self.collected_data
        
        # Demographic summary
        print("📊 SAMPLE DEMOGRAPHICS:")
        print("-" * 30)
        print(f"Total Participants: {len(data)}")
        print(f"Age Distribution: {data['age_group'].value_counts().to_dict()}")
        print(f"Education Distribution: {data['education'].value_counts().to_dict()}")
        print(f"Income Distribution: {data['income'].value_counts().to_dict()}")
        
        # Calculate tolerance scores
        attitude_questions = ['Q6', 'Q7', 'Q8', 'Q9', 'Q10']
        behavior_questions = ['Q11', 'Q12', 'Q13']
        
        data['attitude_score'] = data[attitude_questions].mean(axis=1)
        data['behavior_score'] = data[behavior_questions].mean(axis=1)
        data['overall_tolerance'] = (data['attitude_score'] + data['behavior_score']) / 2
        
        print(f"\n📈 TOLERANCE SCORES:")
        print("-" * 30)
        print(f"Average Attitude Score: {data['attitude_score'].mean():.2f} (SD: {data['attitude_score'].std():.2f})")
        print(f"Average Behavior Score: {data['behavior_score'].mean():.2f} (SD: {data['behavior_score'].std():.2f})")
        print(f"Overall Tolerance Score: {data['overall_tolerance'].mean():.2f} (SD: {data['overall_tolerance'].std():.2f})")
        
        return data
    
    def visualize_sampling_results(self):
        """
        Create visualizations of sampling results and tolerance scores
        """
        if not hasattr(self, 'collected_data'):
            print("No data to visualize. Run collect_data() first.")
            return
        
        data = self.collected_data
        
        # Create figure with subplots
        fig, axes = plt.subplots(2, 2, figsize=(15, 12))
        fig.suptitle('Racial Tolerance Survey: Sampling Results and Initial Findings', 
                    fontsize=16, fontweight='bold')
        
        # Colors for consistent styling
        colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7', '#DDA0DD']
        
        # Plot 1: Demographic distribution of sample
        age_counts = data['age_group'].value_counts()
        axes[0, 0].bar(age_counts.index, age_counts.values, color=colors[:3])
        axes[0, 0].set_title('Sample Age Distribution', fontweight='bold')
        axes[0, 0].set_xlabel('Age Group')
        axes[0, 0].set_ylabel('Number of Participants')
        
        # Add value labels on bars
        for i, v in enumerate(age_counts.values):
            axes[0, 0].text(i, v + 0.1, str(v), ha='center', va='bottom')
        
        # Plot 2: Tolerance scores by age group
        age_tolerance = data.groupby('age_group')['overall_tolerance'].mean()
        bars = axes[0, 1].bar(age_tolerance.index, age_tolerance.values, color=colors[:3])
        axes[0, 1].set_title('Average Tolerance Score by Age Group', fontweight='bold')
        axes[0, 1].set_xlabel('Age Group')
        axes[0, 1].set_ylabel('Average Tolerance Score (1-5)')
        axes[0, 1].set_ylim(1, 5)
        
        # Add value labels on bars
        for bar in bars:
            height = bar.get_height()
            axes[0, 1].text(bar.get_x() + bar.get_width()/2., height + 0.1,
                           f'{height:.2f}', ha='center', va='bottom')
        
        # Plot 3: Tolerance scores by education
        edu_tolerance = data.groupby('education')['overall_tolerance'].mean().sort_values()
        bars_edu = axes[1, 0].barh(list(edu_tolerance.index), edu_tolerance.values, color=colors[3:7])
        axes[1, 0].set_title('Average Tolerance Score by Education', fontweight='bold')
        axes[1, 0].set_xlabel('Average Tolerance Score (1-5)')
        axes[1, 0].set_xlim(1, 5)
        
        # Add value labels on bars
        for i, (edu, score) in enumerate(edu_tolerance.items()):
            axes[1, 0].text(score + 0.05, i, f'{score:.2f}', va='center')
        
        # Plot 4: Attitude vs Behavior scores
        scatter = axes[1, 1].scatter(data['attitude_score'], data['behavior_score'], 
                          alpha=0.7, s=60, color=colors[2])
        axes[1, 1].set_title('Attitude vs Behavior Scores', fontweight='bold')
        axes[1, 1].set_xlabel('Attitude Score (1-5)')
        axes[1, 1].set_ylabel('Behavior Score (1-5)')
        axes[1, 1].set_xlim(1, 5)
        axes[1, 1].set_ylim(1, 5)
        axes[1, 1].grid(True, alpha=0.3)
        
        # Add correlation line
        if len(data) > 1:
            z = np.polyfit(data['attitude_score'], data['behavior_score'], 1)
            p = np.poly1d(z)
            x_range = np.linspace(1, 5, 100)
            axes[1, 1].plot(x_range, p(x_range), "r--", alpha=0.8, label='Trend line')
            axes[1, 1].legend()
        
        plt.tight_layout()
        
        # Save the figure with error handling
        try:
            plt.savefig('outputs/sampling_implementation_results.png', dpi=300, bbox_inches='tight')
            print("✓ Visualizations saved to outputs/sampling_implementation_results.png")
        except Exception as e:
            print(f"⚠️ Could not save visualization: {e}")
            print("Displaying plot instead...")
        
        plt.show()
        
        # Print correlation if we have enough data
        if len(data) > 1:
            correlation = data['attitude_score'].corr(data['behavior_score'])
            print(f"📊 Attitude-Behavior Correlation: r = {correlation:.3f}")

# Execute the sampling implementation
def implement_sampling_study():
    """
    Main function to implement the complete sampling study
    """
    print("🎯 IMPLEMENTING STRATIFIED SAMPLING FOR RACIAL TOLERANCE SURVEY")
    print("="*70)
    
    try:
        # Initialize sampling class
        sampling_study = RacialToleranceSampling()
        
        # Collect data from 15 participants (you can increase this number)
        collected_data = sampling_study.collect_data(sample_size=15)
        
        # Analyze sample characteristics
        analyzed_data = sampling_study.analyze_sample_characteristics()
        
        # Create visualizations
        sampling_study.visualize_sampling_results()
        
        # Save collected data
        try:
            collected_data.to_csv('data/primary_survey_data.csv', index=False)
            print(f"\n💾 Survey data saved to: data/primary_survey_data.csv")
        except Exception as e:
            print(f"⚠️ Could not save data file: {e}")
        
        # Display sample of collected data
        print("\n📋 SAMPLE OF COLLECTED DATA:")
        print("-" * 50)
        display_cols = ['id', 'age_group', 'education', 'income', 'attitude_score', 'behavior_score', 'overall_tolerance']
        
        # Create a clean display without index
        display_data = analyzed_data[display_cols].head(8).round(2)
        for col in ['attitude_score', 'behavior_score', 'overall_tolerance']:
            display_data[col] = display_data[col].round(2)
        
        print(display_data.to_string(index=False))
        
        # Print summary statistics
        print(f"\n📈 SUMMARY STATISTICS:")
        print("-" * 30)
        print(f"Total participants: {len(analyzed_data)}")
        print(f"Age range: {analyzed_data['age_group'].unique()}")
        print(f"Education levels: {len(analyzed_data['education'].unique())}")
        print(f"Income brackets: {len(analyzed_data['income'].unique())}")
        print(f"Overall tolerance range: {analyzed_data['overall_tolerance'].min():.2f} - {analyzed_data['overall_tolerance'].max():.2f}")
        
        return analyzed_data
        
    except Exception as e:
        print(f"❌ Error in sampling implementation: {e}")
        print("Please check the error message above and try again.")
        return None

# Run the sampling implementation
if __name__ == "__main__":
    survey_data = implement_sampling_study()
    
    if survey_data is not None:
        print(f"\n{'='*70}")
        print("🎉 SAMPLING STUDY COMPLETED SUCCESSFULLY!")
        print(f"{'='*70}")
        print("Generated outputs:")
        print("✓ outputs/sampling_implementation_results.png - Sampling visualizations")
        print("✓ data/primary_survey_data.csv - Collected survey data")
        print(f"✓ {len(survey_data)} complete survey responses")
    else:
        print(f"\n❌ Sampling study failed. Please check the implementation.")
```

# Evaluation of Sampling Error: Perceived Racial Tolerance Project
# 1. Potential Sources of Sampling Error
# Sampling error occurs when the sample selected is not perfectly representative 
# of the target population, leading to results that differ from what would be found 
# if the entire population were surveyed.

# Source 1: Coverage Error (Frame Error)

# Description: This occurs when the sampling frame (the list from which the sample is drawn) 
# does not adequately cover the entire target population. For example, if the study uses a
# n online survey distributed via social media or email lists, it systematically excludes 
# populations with limited internet access, such as the elderly, low-income individuals, 
# and people in rural areas with poor connectivity.

# Effect on Results: The results would over-represent the views of digitally 
# connected demographics and under-represent the experiences of digitally disconnected groups. This could lead to an over- or under-estimation of the overall level of perceived racial tolerance.

# Source 2: Non-Response Bias

# Description: This happens when individuals selected for the sample choose 
# not to participate. In a sensitive topic like racial tolerance, 
# individuals with strong views (either highly tolerant or highly intolerant) 
# may be more or less likely to respond. For instance, those who are racially 
# intolerant might avoid the survey to not reveal their views, or conversely, 
# those passionate about social justice might be over-eager to respond.

# Effect on Results: If individuals with intolerant views systematically avoid the survey, 
# the results would show a significantly higher level of perceived racial tolerance than 
# actually exists in the population. The "perception" would be skewed toward the 
# perceptions of those willing to talk about race.

# Source 3: Sampling Bias (Selection Bias)

# Description: This arises from the sampling method itself being non-random. 
# If the study relies on a convenience sample 
# (e.g., surveying students on a university campus or people in a single, 
# politically homogeneous neighborhood), the sample will not represent the 
# broader population's geographic, educational, or ideological diversity.

# Effect on Results: A sample from a liberal arts college campus would likely yield a 
# very different perception of racial tolerance compared to a sample from a more 
# politically or demographically heterogeneous community. The results become 
# largely ungeneralizable beyond the specific group sampled.


2. Minimizing Sampling Errors in Future Studies
To Minimize Coverage Error:

Use a mixed-mode approach. Combine online surveys with telephone surveys (including cell phones) and mailed questionnaires to reach populations with varying levels of internet access.

If the target population is a specific city or country, use a master list like a census tract or a list of all residential addresses as the sampling frame instead of an online list.

To Minimize Non-Response Bias:

Employ follow-up procedures, such as sending reminders and making multiple contact attempts at different times/days.

Offer incentives for participation (e.g., gift cards, entry into a lottery) to encourage a broader range of people to respond.

Ensure strong confidentiality and anonymity assurances to make respondents feel safe in answering sensitive questions truthfully.

To Minimize Sampling Bias:

Implement a Probability Sampling method. The gold standard is Simple Random Sampling from a comprehensive sampling frame. If that's not feasible, Stratified Random Sampling can ensure key subgroups (e.g., different racial groups, age brackets, income levels) are proportionally represented in the sample.



3. Calculation of Margin of Error
The Margin of Error (MOE) quantifies the uncertainty caused by surveying a sample rather than the whole population. It provides a range (confidence interval) within which the true population value is likely to fall.

Calculation:

The formula for the margin of error at a 95% confidence level is:
MOE = 1.96 * √[ p * (1 - p) / n ]

Where:

1.96 is the z-score for a 95% confidence level.

p is the sample proportion (expressed as a decimal). If not specified, we use p = 0.5 (50%) as it provides the maximum, most conservative margin of error.

n is the sample size.

Example Calculation for the Project:

Let's assume the Perceived Racial Tolerance Project collected a random sample of n = 1,000 people. We will calculate the maximum margin of error.

MOE = 1.96 * √[ (0.5) * (1 - 0.5) / 1000 ]

MOE = 1.96 * √[ 0.25 / 1000 ]

MOE = 1.96 * √0.00025

MOE = 1.96 * 0.01581

MOE ≈ 0.0310 or 3.1%

Interpretation:
For a sample size of 1,000, the margin of error is approximately ±3.1% at a 95% confidence level. This means if the survey finds that 70% of the sample perceives a high level of racial tolerance, we can be 95% confident that the true proportion in the entire population lies between 66.9% and 73.1% (70% ± 3.1%).

(Note: This calculation assumes a simple random sample was used. The presence of the other sampling errors (coverage, non-response) means the true error might be larger than this calculated margin.)
